# 화공컴퓨터모델링 팀프로젝트 — 수치해석 (Python)

**문제.** 반응 평형 조성 x를 비선형 방정식의 근으로 구한다.

$$f(x) = \dfrac{4+x}{(42-2x)^2\,(28-x)} - 0.016 = 0$$

구간 [0, 20]에서 참 근은 **x ≈ 15.923** 하나뿐이다. (`f(15.923) ≈ 0`)



## 1. 가위치법 (False Position)

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

Es = 0.0001   # 기존 0.5. 가위치법이 참 근(15.923)에 도달하도록 더 엄격하게 설정.
x_r0=0 #임의로 설정
x_l=0
x_l1=x_l
x_u=20
x_u1=x_u
i = 0
fx_l=(4+x_l)/((42-2*x_l)**2)/(28-x_l)-0.016
fx_u=(4+x_u)/((42-2*x_u)**2)/(28-x_u)-0.016
x_r=(x_l*fx_u-x_u*fx_l)/(fx_u-fx_l)
Ea = (x_r-x_r0)/x_r * 100
while Ea >= Es :
    x_r=(x_l*fx_u-x_u*fx_l)/(fx_u-fx_l)
    fx_r=(4+x_r)/((42-2*x_r)**2)/(28-x_r)-0.016
    #print("f(x_r)=",fx_r)
    if fx_r < 0: # fx_r < 0 일 때
        x_l = x_r
        #print ("x_r이 x_l이 되었습니다.",x_l)
        fx_l=(4+x_l)/((42-2*x_l)**2)/(28-x_l)-0.016
    else:
        x_u = x_r
        #print ("x_r이 x_u가 되었습니다.",x_u)
        fx_u=(4+x_u)/((42-2*x_u)**2)/(28-x_u)-0.016
    Ea = (x_r-x_r0)/x_r * 100
    i = i+1
    #print("Ea =",Ea, ", i=",i)
    Et = abs(x_r-15.923)/x_r*100
    x_r0=x_r
    #print("---")
print("False-Position")
print("Initial Approximation x_l:",x_l1)
print("Initial Approximation x_u:", x_u1)
print("Es:", Es)
print("Iteration:", i)
print("The root of the equation is:", x_r)
print("Ea:", Ea,"%")
```

## 2. 이분법 (Bisection)

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

x_l=0
x_l1=0
x_u=20
x_u1=20
x_r0=0
x_r=(x_l+x_u)/2
Ea = (x_r-x_r0)/x_r * 100
Es = 0.0001   # 기존 0.5. 네 방법의 근을 일치시키기 위해 더 엄격하게 설정.
i=0
while abs(Ea) >= Es :
    x_r=(x_l+x_u)/2
    fx_r=(4+x_r)/((42-2*x_r)**2)/(28-x_r)-0.016
    #print("f(x_r) =",fx_r)
    if fx_r < 0: # fx_r < 0 일 때
        x_l = x_r
        #print ("x_r equals x_l.",x_l)
    else:
        x_u = x_r
        #print ("x_re equals x_u.",x_u)
    i=i+1
    Ea = (x_r-x_r0)/x_r * 100
    x_r0=x_r
    #print("i is", i,"," , "Ea is",Ea)
    #print("---")
print("Bisection")
print("Initial Approximation x_l:",x_l1)
print("Initial Approximation x_u:", x_u1)
print("Es:", Es)
print("Iteration:", i)
print("The root of the equation is:", x_r)
print("Ea:", Ea)
```

## 3. 뉴턴-랩슨법 (Newton-Raphson)



```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

x_r=18
x_r0 = x_r
x_r1 = x_r
i=0
Ea = 100
Es = 0.0001   # 기존 0.5. 네 방법의 근을 일치시키기 위해 더 엄격하게 설정.
x_r0=0
while Ea >= Es :
    i = i+1
    fx_r=(4+x_r)/((42-2*x_r)**2)/(28-x_r)-0.016
    dfx_r= -(-448 - 8*x_r + x_r**2)/(2 * ((21 - x_r)**3)*(-28 + x_r)**2)
    x_r = x_r-fx_r/dfx_r
    #print("x_r =",x_r)
    Ea = abs((x_r-x_r0)/x_r*100)
    x_r0 = x_r
    #print("i =",i, "x_r =", x_r,"Ea =", Ea)
print("Newton-Raphson")
print("Initial Approximation x:", x_r1)
print("Es:", Es)
print("Iteration:", i)
print("The root of the equation is:", x_r)
print("Ea:", Ea,"%")
```

## 4. 고정점 반복법 (Fixed-Point Iteration)


```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

# x = g(x) 형태로 변형: (42-2x)^2 = (4+x)/(0.016*(28-x))
#   -> 42-2x = sqrt((4+x)/(0.016*(28-x)))  (근 부근에서 42-2x > 0)
#   -> x = 21 - 0.5*sqrt((4+x)/(0.016*(28-x)))
x_r=18
x_r1 = x_r
i=0
Ea = 100
Es = 0.0001   # 참 근(15.923)으로 수렴하도록 설정.
x_r0=0
while Ea >= Es :
    i = i+1
    x_r = 21 - 0.5*((4+x_r)/(0.016*(28-x_r)))**0.5
    #print("x_r =",x_r)
    Ea = abs((x_r-x_r0)/x_r*100)
    x_r0 = x_r
    #print("i =",i, "x_r =", x_r,"Ea =", Ea)
print("Fixed-Point Iteration")
print("Initial Approximation x:", x_r1)
print("Es:", Es)
print("Iteration:", i)
print("The root of the equation is:", x_r)
print("Ea:", Ea,"%")
```
